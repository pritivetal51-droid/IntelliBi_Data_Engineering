```python
%sql
COPY INTO intellibi_catalog.intellibi_bronze.rawMonthlySales
FROM (
    SELECT sale_id, product, category, quantity, price, sale_date, region, 
     current_timestamp() AS ingest_ts 
    FROM '/Volumes/intellibi_catalog/intellibi_schema/intellibi_landing_files/src_files/monthly_sales/'
     )
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'false')
```
